Data Preprocessing - test train split and normalization

```python
# Dataset Path
base_dir = 'plantvillage dataset/color'
```

```python
# Image Parameters
img_size = 224 #we are resizing all images 224x224 as it is the standard and saves memmory
batch_size = 32 # indicates number of images that are used for training at once
```

ImageDataGenerator is a Keras utility used for loading images in batches and augmenting them for pre processing. Keras validation split can only produce two subsets so if we to do 80/10/10 split for obtaining a global test set then we need to manually divide the dataset into three subsets using pandas.

```python
# Image Data Generators
data_gen = ImageDataGenerator(
    rescale=1./255, # used to get all pixel values to lie bewtween 0 and 1 ie a gaussian distribution(normalisation)
    validation_split=0.2  # Use 20% of data for validation and 80% for testing
)
```

flow_from_directory looks into the dataset and assigns class from the folder names and groups images into batches. useful in multiclass classification

```python
# Train Generator
train_generator = data_gen.flow_from_directory(
    base_dir,
    target_size=(img_size, img_size), # setting image size 224x224
    batch_size=batch_size,
    subset='training',
    class_mode='categorical' #for one hot encoding.
)
```

```python
# Validation Generator
validation_generator = data_gen.flow_from_directory(
    base_dir,
    target_size=(img_size, img_size),
    batch_size=batch_size,
    subset='validation',
    class_mode='categorical'
)
```

Sequential is a class in the models module of keras. It is used to build a CNN in a sequential manner ie one layer after another. model.add is used to specify what the next layer is in the sequence.

First layer performs convolution. we used 32 filters in the first layer where each filter is a 3x3 kernel which slides over the entire image producing a feature map of size 222x222. since 32 filters are used therefore we get 32 features maps of size 222x222.

input_shape is used to specify the size of input image.

Next layer is a max pooling layer (2,2) specifies we are seeing 2x2 block on the feature map and in this block we take the max value. so every 2x2 block on the feature map returns its max value. also stride for max pooling is taken as 2 so it moves across 2 coulmmns/2 rows each time. hence size of the feature map is reduced by half and thus computation is decreased. so the output of this will be a feature map of size 111x111x32

third layer is again convolution but with 64 filters. output of this will be 109x109x64.

fourth layer is max pooling. output of this layer will be 54x54x64

fifth layer is used to flatten as then we use a dense layer and it can take only 1D vectors as its input.

sixth layer is a dense layer ie a fully connected layer with 256 neurons. the output of thus layer will be 256 values.

7th layer is the output layer. train_generator.num_classes returns the number of classes in the  train_generator dataset. it returns 38(1 class for healthy leaf and 37 classes for diseases). activation function use dis softmax as it is multiclass problem.

model.summary summarises which layers are present in the model.


```python
# Model Definition
model = models.Sequential()

model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(img_size, img_size, 3)))
model.add(layers.MaxPooling2D(2, 2))

model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D(2, 2))


model.add(layers.Flatten())
model.add(layers.Dense(256, activation='relu'))
model.add(layers.Dense(train_generator.num_classes, activation='softmax'))
```

```python
# model summary
model.summary()
```

```python
# Compile the Model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```
